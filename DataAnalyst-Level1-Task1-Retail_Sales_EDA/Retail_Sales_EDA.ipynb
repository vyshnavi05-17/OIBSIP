## 3. Time Series Analysis

Time series analysis is performed to understand how retail sales change over time. Monthly and quarterly sales trends are analyzed to identify periods of high and low sales activity.

```python
monthly_sales = df.groupby(
    df['Date'].dt.to_period('M')
)['Total Amount'].sum()

monthly_sales
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Observation

The monthly sales trend shows fluctuations in sales across the analyzed period from 2023 to January 2024. The highest monthly sales were recorded in May 2023, with total sales of 53,150, while the lowest sales were recorded in January 2024, with total sales of 1,530.

The significant variation in monthly sales indicates changes in customer purchasing activity over time. The business can use these trends to improve inventory planning and schedule promotional campaigns during periods of higher demand.

### 3.2 Quarterly Sales Trend

Quarterly sales analysis provides a broader view of business performance by grouping monthly transactions into four quarters.

```python
quarterly_sales = df.groupby(
    ['Year', 'Quarter']
)['Total Amount'].sum()

quarterly_sales
```

```python
quarterly_labels = [
    f'{year} Q{quarter}'
    for year, quarter in quarterly_sales.index
]

plt.figure(figsize=(10, 6))

plt.plot(
    quarterly_labels,
    quarterly_sales.values,
    marker='o'
)

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```

### Observation

Quarterly sales performance varies across the analyzed period. The highest sales were recorded in Q4 2023, with total sales of 126,190, while the lowest recorded value was 1,530 in Q1 2024.

However, Q1 2024 contains only January data in the dataset and therefore should not be directly compared with the complete quarters of 2023. Among the complete quarters, Q4 2023 recorded the highest sales.

The quarterly analysis can help management identify stronger sales periods and improve inventory planning, marketing activities, and sales targets.

```python
highest_month = monthly_sales.idxmax()
highest_month_value = monthly_sales.max()

lowest_month = monthly_sales.idxmin()
lowest_month_value = monthly_sales.min()

print("Highest sales month:", highest_month)
print("Highest sales:", highest_month_value)

print("\nLowest sales month:", lowest_month)
print("Lowest sales:", lowest_month_value)
```

```python
highest_quarter = quarterly_sales.idxmax()
highest_quarter_value = quarterly_sales.max()

lowest_quarter = quarterly_sales.idxmin()
lowest_quarter_value = quarterly_sales.min()

print("Highest sales quarter:", highest_quarter)
print("Highest sales:", highest_quarter_value)

print("\nLowest sales quarter:", lowest_quarter)
print("Lowest sales:", lowest_quarter_value)
```

## 4. Customer Demographic Analysis

Customer demographic analysis is performed to understand the characteristics of the retail customer base. Age-group and gender distributions are analyzed to identify important customer segments.

```python
bins = [0, 18, 25, 35, 45, 55, 65, 100]

labels = [
    'Under 18',
    '18-25',
    '26-35',
    '36-45',
    '46-55',
    '56-65',
    '65+'
]

df['Age Group'] = pd.cut(
    df['Age'],
    bins=bins,
    labels=labels
)
```

```python
age_group_counts = df['Age Group'].value_counts().sort_index()

age_group_counts
```

```python
plt.figure(figsize=(10, 6))

age_group_counts.plot(kind='bar')

plt.title('Customer Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
age_group_counts
```

### Observation

The customer age distribution shows that the 46–55 age group is the largest customer segment, accounting for 229 customers (22.9%) of the dataset.

The 26–35 and 36–45 age groups account for 205 (20.5%) and 202 (20.2%) customers respectively, while the 56–65 group represents 195 customers (19.5%). Together, these groups indicate that customers between 26 and 65 years make up the majority of the customer base.

The 18–25 age group represents 148 customers (14.8%), while the Under 18 group has only 21 customers (2.1%). No customers were recorded in the 65+ category.

### Business Insight

The business appears to have a strong customer base among adults aged 26–65, particularly the 46–55 segment. Marketing campaigns, product recommendations, and loyalty programs could therefore be tailored toward these high-representation age groups.

## 5. Product Analysis

Product analysis is performed to identify the best-performing product categories based on sales quantity and revenue.

```python
category_quantity = df.groupby(
    'Product Category'
)['Quantity'].sum().sort_values(ascending=False)

category_quantity
```

```python
plt.figure(figsize=(9, 6))

category_quantity.plot(kind='bar')

plt.title('Quantity Sold by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Quantity Sold')

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
category_revenue = df.groupby(
    'Product Category'
)['Total Amount'].sum().sort_values(ascending=False)

category_revenue
```

```python
product_analysis = df.groupby('Product Category').agg(
    Quantity_Sold=('Quantity', 'sum'),
    Revenue=('Total Amount', 'sum')
).reset_index()

product_analysis
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=product_analysis,
    x='Quantity_Sold',
    y='Revenue',
    s=200
)

for i, row in product_analysis.iterrows():
    plt.text(
        row['Quantity_Sold'],
        row['Revenue'],
        row['Product Category'],
        fontsize=10,
        ha='left',
        va='bottom'
    )

plt.title('Quantity Sold vs Revenue by Product Category')
plt.xlabel('Total Quantity Sold')
plt.ylabel('Total Revenue')

plt.tight_layout()
plt.show()
```

### Observation

The scatter plot shows the relationship between quantity sold and total revenue across product categories. Clothing has the highest quantity sold, with approximately 895 units, but Electronics generates the highest revenue despite having a lower sales quantity.

This indicates that Electronics products have a higher revenue contribution per unit compared with Clothing. Beauty has both the lowest quantity sold and the lowest total revenue among the three categories.

### Business Insight

The business should not evaluate product performance based only on sales volume. Electronics appears to generate strong revenue with fewer units sold, suggesting that higher-priced products can contribute significantly to overall revenue. The business could focus on maintaining Electronics inventory while using targeted promotions to increase Beauty sales.

## 6. Correlation Analysis

Correlation analysis is used to examine the relationships between numerical variables in the retail sales dataset. A heatmap is used to visually identify positive and negative relationships between variables.

```python
correlation_matrix = df[
    ['Age', 'Quantity', 'Price per Unit', 'Total Amount']
].corr()

correlation_matrix
```

```python
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables')

plt.tight_layout()
plt.show()
```

### Observation

The correlation heatmap shows that Price per Unit and Total Amount have a strong positive correlation of 0.85. This indicates that transactions with higher-priced products generally generate higher total transaction amounts.

Quantity and Total Amount show a moderate positive correlation of 0.37, suggesting that purchasing more units tends to increase the total transaction value, although the relationship is not very strong.

Age has very weak correlations with Total Amount (-0.06), Quantity (-0.02), and Price per Unit (-0.04). This suggests that customer age has little linear relationship with purchasing quantity, product price, or transaction value in this dataset.

### Business Insight

Product pricing appears to be a stronger driver of transaction value than customer age. The business should therefore pay attention to product pricing and high-value product categories when developing revenue-growth strategies.

## 7. Transaction Value Distribution

Analyzing the distribution of transaction values helps identify typical customer spending levels and determine whether the dataset contains unusually high-value transactions.

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df['Total Amount'],
    bins=20,
    edgecolor='black'
)

plt.title('Distribution of Transaction Amounts')
plt.xlabel('Total Transaction Amount')
plt.ylabel('Number of Transactions')

plt.tight_layout()
plt.show()
```

```python
mean_amount = df['Total Amount'].mean()
median_amount = df['Total Amount'].median()

print("Mean Transaction Amount:", round(mean_amount, 2))
print("Median Transaction Amount:", round(median_amount, 2))
```

```python
mean_amount = df['Total Amount'].mean()
median_amount = df['Total Amount'].median()

print("Mean Transaction Amount:", round(mean_amount, 2))
print("Median Transaction Amount:", round(median_amount, 2))
```

### Observation

The distribution of transaction amounts is strongly right-skewed. Most transactions are concentrated in the lower transaction-value range, while a smaller number of transactions have substantially higher values, extending toward approximately 2,000.

This indicates that high-value transactions occur less frequently than low-value transactions but may contribute significantly to overall revenue. Therefore, the business should consider both average and median transaction values when evaluating customer spending behaviour.

### Business Insight

The business could identify and target customers associated with higher-value transactions through loyalty programs, personalized recommendations, and premium product offers. At the same time, promotional strategies can be used to increase the value of lower-value purchases through product bundles and cross-selling.

# 8. Conclusion

The Exploratory Data Analysis of the retail sales dataset provided several useful insights into sales trends, customer demographics, product performance, and transaction behaviour.

The time series analysis showed that sales fluctuated across the analyzed period, with May 2023 recording the highest monthly sales of 53,150. Among the complete quarters, Q4 2023 recorded the highest sales of 126,190.

Customer demographic analysis showed that the 46–55 age group was the largest customer segment, representing 22.9% of the dataset. The gender distribution was relatively balanced, with female customers accounting for 51% and male customers accounting for 49%.

Product analysis revealed an interesting difference between sales volume and revenue. Clothing had the highest quantity sold, while Electronics generated the highest revenue. This indicates that higher sales volume does not necessarily result in higher revenue.

The correlation analysis showed a strong positive relationship of 0.85 between Price per Unit and Total Amount. Quantity also had a moderate positive relationship with Total Amount (0.37), while Age showed very weak relationships with the other numerical variables.

The transaction-value distribution was strongly right-skewed, indicating that most transactions were relatively low-value while a smaller number of high-value transactions contributed significantly to overall revenue.

# 9. Actionable Business Recommendations

### 1. Focus on high-value Electronics products

Electronics generated the highest revenue despite having a lower sales quantity than Clothing. The business should maintain sufficient inventory of high-performing Electronics products and consider targeted promotions for these products to maximize revenue.

### 2. Target the 26–65 customer segments

Customers aged 26–65 represent the majority of the customer base, with the 46–55 group being the largest segment. The business should develop targeted marketing campaigns, loyalty programs, and personalized product recommendations for these customer groups.

### 3. Increase the value of low-value transactions

The transaction distribution shows that most purchases are concentrated in the lower-value range. The business could increase average transaction value through product bundles, cross-selling, minimum-purchase discounts, and complementary product recommendations.

### 4. Use seasonal sales patterns for planning

Q4 2023 recorded the highest complete-quarter sales. The business should analyze successful strategies used during strong sales periods and prepare inventory and promotional campaigns ahead of similar high-demand periods.

### 5. Prioritize pricing and product mix

The strong correlation between Price per Unit and Total Amount (0.85) suggests that product price has a major influence on transaction value. The business should optimize its product mix by balancing high-value products with products that generate strong sales volume.
